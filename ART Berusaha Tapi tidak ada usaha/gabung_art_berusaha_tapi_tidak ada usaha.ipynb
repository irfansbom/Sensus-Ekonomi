```python
import pandas as pd
from pathlib import Path
from datetime import datetime
from openpyxl import load_workbook
FORMAT_TANGGAL_JAM = "%d-%m-%Y_%H-%M-%S"
tanggal_jam = datetime.now().strftime(FORMAT_TANGGAL_JAM)
```

```python
folder = Path("./palembang")
files = list(
    folder.glob(
        "sqllab_art_berusaha_tapi_keluarga_tidak_ada_usaha*.xlsx"
    )
)

if not files:
    raise FileNotFoundError("Tidak ada file Excel di folder palembang")


print(f"Ditemukan {len(files)} file:")
for f in files:
    print(" -", f.name)

# Baca semua file lalu gabungkan
list_df = []
for f in files:
    df = pd.read_excel(f, dtype=str)
    df["source_file"] = (
        f.name
    )  # opsional: kolom penanda asal file, bisa dihapus kalau tidak perlu
    list_df.append(df)

df_gabungan = pd.concat(list_df, ignore_index=True)

print(f"\nTotal baris setelah digabung: {len(df_gabungan)}")
df_gabungan
```

```python
df_gabungan.to_excel(f"{folder}/art_berusaha_tapi_tidak_ada_usaha_{tanggal_jam}.xlsx", index=False)
```
