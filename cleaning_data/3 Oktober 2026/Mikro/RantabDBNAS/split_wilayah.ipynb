```python
import pandas as pd
from pathlib import Path
#
folder = Path(".")  # <-- ganti dengan folder CSV Anda


kolom_str = ["assignment_id", "level_2_code", "kategori", "kode_kbli", "nama_usaha"]

kolom_int = [
    "metrik_tenaga_kerja",
    "metrik_upah_gaji",
    "metrik_biaya_produksi",
    "metrik_biaya_pembelian",
    "metrik_biaya_operasional",
    "metrik_biaya_non_operasional",
    "metrik_total_pengeluaran",
    "metrik_pendapatan_barang_jasa",
    "metrik_pendapatan_lainnya",
    "metrik_nilai_tambah",
    "metrik_total_aset",
    "metrik_output",
]

kolom_real = [
    "rasio_biaya_pembelian_barang_terhadap_omzet",
    "rasio_ntb",
    "produktivitas",
]

# Tidak disebut: kategori/flag, diperlakukan string
kolom_lain_str = [
    "reklasifikasi_skala_usaha",
    "flag_rasio_1_output_aset",
    "flag_rasio_2_upah_ntb",
    "flag_rasio_3_ntb_output",
    "flag_kategori_P_dan_U",
]
```

```python
# Baca semua sebagai string dulu, konversi sesudahnya (aman dari error tipe saat ada nilai kosong/aneh)
dtype_baca = {c: str for c in kolom_str + kolom_int + kolom_real + kolom_lain_str}

frames = []
for f in sorted(folder.glob("*.csv")):
    d = pd.read_csv(f, dtype=dtype_baca, encoding="utf-8-sig", low_memory=False)
    d.columns = d.columns.astype(str).str.strip()
    d["sumber_file"] = f.name
    frames.append(d)
    print(f"{f.name}: {d.shape}")

df = pd.concat(frames, ignore_index=True)

# Buang baris header yang ikut terbaca sebagai data
df = df[df["assignment_id"] != "assignment_id"]

# String: strip, nilai kosong jadi None
for c in kolom_str + kolom_lain_str:
    df[c] = (
        df[c]
        .astype("string")
        .str.strip()
        .replace({"": pd.NA, "nan": pd.NA, "None": pd.NA})
    )

# Integer: pakai Int64 (nullable) supaya NaN tidak memaksa jadi float
for c in kolom_int:
    df[c] = pd.to_numeric(df[c], errors="coerce").round().astype("Int64")

# Real: float64
for c in kolom_real:
    df[c] = pd.to_numeric(df[c], errors="coerce").astype("float64")

df = df.drop_duplicates().reset_index(drop=True)

print(df.shape)
print(df.dtypes)
df.head()
```

```python
BASE_URL = (
    "https://fasih-sm.bps.go.id/app/assignment/fd68e454-ba45-4b85-8205-f3bf777ded24/"
)

# 1. Kolom link, ditaruh paling kiri
df.insert(0, "link", BASE_URL + df["assignment_id"].astype("string"))

# 2. Dua kolom di ujung kanan
df["status_penyelesaian"] = False  # nanti jadi checkbox di Excel
df["keterangan"] = ""  # kosong

# 3. Pastikan dua kolom baru benar-benar paling ujung (sumber_file digeser sebelum keduanya)
akhir = ["status_penyelesaian", "keterangan"]
df = df[[c for c in df.columns if c not in akhir] + akhir]

df.head()
```

```python
# SPLIT Data
import re
from pathlib import Path
import pandas as pd

out_dir = Path("output_excel")
out_dir.mkdir(exist_ok=True)


def nama_aman(teks, maks=31):
    t = re.sub(r"[\[\]:*?/\\]", "_", str(teks)).strip()
    return (t or "KOSONG")[:maks]


kolom_export = [c for c in df.columns if c != "sumber_file"]
kolom_teks = {"link", "assignment_id", "level_2_code", "kode_kbli"}
idx_status = kolom_export.index("status_penyelesaian")

for lv2, df_lv2 in df.groupby("level_2_code", dropna=False):
    path = out_dir / f"{nama_aman(lv2, 100)}.xlsx"
    sheet_terpakai = set()

    # strings_to_urls=False: link ditulis sebagai teks biasa (Excel membatasi 65.530 hyperlink per sheet)
    with pd.ExcelWriter(
        path, engine="xlsxwriter", engine_kwargs={"options": {"strings_to_urls": False}}
    ) as writer:
        wb = writer.book
        fmt_header = wb.add_format({"bold": True, "bg_color": "#D9E1F2", "border": 1})
        fmt_teks = wb.add_format({"num_format": "@"})

        for kat, d in df_lv2.groupby("kategori", dropna=False):
            nama = nama_aman(kat)
            asli, n = nama, 1
            while nama.lower() in sheet_terpakai:
                n += 1
                nama = f"{asli[:28]}_{n}"
            sheet_terpakai.add(nama.lower())

            d = d[kolom_export].reset_index(drop=True)
            d.to_excel(writer, sheet_name=nama, index=False)
            ws = writer.sheets[nama]

            for j, col in enumerate(d.columns):
                ws.write(0, j, col, fmt_header)
            ws.freeze_panes(1, 0)
            ws.autofilter(0, 0, len(d), len(d.columns) - 1)

            # Checkbox pada kolom status_penyelesaian (default FALSE / tidak dicentang)
            for i in range(len(d)):
                ws.insert_checkbox(
                    i + 1, idx_status, bool(d.at[i, "status_penyelesaian"])
                )

            for j, col in enumerate(d.columns):
                lebar = max(
                    len(str(col)), d[col].head(1000).astype(str).str.len().max()
                )
                ws.set_column(
                    j, j, min(lebar + 2, 40), fmt_teks if col in kolom_teks else None
                )

    print(
        f'{path.name}: {df_lv2["kategori"].nunique(dropna=False)} sheet, {len(df_lv2)} baris'
    )
```

```python
import sqlite3

DB_PATH = "usaha.db"

df_sql = df.copy()
df_sql["status_penyelesaian"] = df_sql["status_penyelesaian"].astype(
    int
)  # SQLite tidak punya boolean: 0/1

# Tipe kolom eksplisit di SQLite
dtype_sql = {}
for c in df_sql.columns:
    if c in kolom_int or c == "status_penyelesaian":
        dtype_sql[c] = "INTEGER"
    elif c in kolom_real:
        dtype_sql[c] = "REAL"
    else:
        dtype_sql[c] = "TEXT"

with sqlite3.connect(DB_PATH) as con:
    df_sql.to_sql(
        "usaha",
        con,
        if_exists="replace",
        index=False,
        dtype=dtype_sql,
        chunksize=50_000,
    )
    # Index agar filter cepat
    con.execute("CREATE INDEX IF NOT EXISTS idx_lv2 ON usaha(level_2_code)")
    con.execute("CREATE INDEX IF NOT EXISTS idx_kat ON usaha(kategori)")
    con.execute("CREATE INDEX IF NOT EXISTS idx_aid ON usaha(assignment_id)")
    con.commit()

    print(con.execute("SELECT COUNT(*) FROM usaha").fetchone())
```
