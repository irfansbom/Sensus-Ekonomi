```python
import pandas as pd
import math
import os
```

```python
# ==== KONFIGURASI ====
CSV_PATH = "../Data DTSEN.csv"  # path file csv input
KOLOM_NIK = "NIK"  # nama kolom NIK di csv
BATCH_SIZE = 1000  # jumlah NIK per file
OUTPUT_DIR = "sql_cek_nik_output"  # folder output

TEMPLATE = """SELECT assignment_id, assignment_status_alias, nama_dtsen, nik_dtsen, keberadaan_dtsen_label
FROM nested_dtsen
WHERE nik_dtsen IN (
{nik_list}
);
"""
```

```python
os.makedirs(OUTPUT_DIR, exist_ok=True)

df = pd.read_csv(CSV_PATH, dtype=str)
df.head()
niks = df[KOLOM_NIK].dropna().astype(str).str.strip()
niks = niks[niks != ""].tolist()

total = len(niks)
n_batch = math.ceil(total / BATCH_SIZE)

print(f"Total NIK: {total}")
print(f"Jumlah file yang dibuat: {n_batch}")

for i in range(n_batch):
    batch = niks[i * BATCH_SIZE : (i + 1) * BATCH_SIZE]
    nik_list_str = ",\n".join(f"'{nik}'" for nik in batch)
    sql_content = TEMPLATE.format(nik_list=nik_list_str)

    filename = os.path.join(OUTPUT_DIR, f"query_batch_{i+1:02d}.sql")
    with open(filename, "w", encoding="utf-8") as f:
        f.write(sql_content)

    print(f"  -> {filename} ({len(batch)} NIK)")

print("Selesai.")
```
